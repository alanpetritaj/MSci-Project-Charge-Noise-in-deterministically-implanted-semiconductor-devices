# Island 1 Coulomb diamonds

This notebook runs the island model from `Island1_functions.ipynb` and builds up the Coulomb diamonds one step at a time.

Later on, a defect charge switching near the island will shift the diamonds a little, and that shift is the charge noise. To see a shift we first need a clean defect-free picture that we understand.

```python
# pull in the classes and functions (Island1_functions.ipynb has to be in the same folder)
%run Island1_function.ipynb
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch
%matplotlib inline
plt.rcParams.update({'figure.dpi': 110, 'axes.grid': True, 'grid.alpha': .25, 'font.size': 10})

# same colours everywhere, light blue = even n, dark blue = odd n
EVEN_C, ODD_C = '#9ecae1', '#3182bd'
```

## 1. The island and its numbers

$E_C$, $\alpha_g$ and $c = C_S/C_\Sigma$ are put in from the thesis, and $c = 0.5$ is an assumed symmetric island. The rest of the parameters are:

$$C_\Sigma = \frac{e^2}{E_C}, \qquad C_g = \alpha_g C_\Sigma, \qquad C_S = c\,C_\Sigma, \qquad C_D = C_\Sigma - C_S - C_g .$$

Island.centred(n_ref=6) puts the $n = 6$ diamond at $V_G = 0$.

$n$ is not the real number of electrons on the island. The code only keeps the 16 levels near the Fermi energy, and $n$ counts the electrons added to them from an arbitrary starting point. The electrons below that never change, so they just get absorbed into the background charge $q_0$. Choosing 6 is just arbitrary.

```python
# build the island, n=6 diamond sits at V_G = 0
isl = Island.centred(n_ref=6)

print(f"E_C = {isl.p.E_C/meV:.2f} meV,  alpha_g = {isl.p.alpha_g:.2f},  c = {isl.p.c_S:.2f},  dE = {isl.p.dE/meV:.2f} meV,  g = {isl.p.g}")
print(f"C_sigma = {isl.C_sigma/aF:.2f} aF,  C_g = {isl.C_g/aF:.2f} aF,  C_S = {isl.C_S/aF:.2f} aF,  C_D = {isl.C_D/aF:.2f} aF")
```

## 2. The ladder of levels

On top of the charging energy the electrons sit in single-particle levels $\Delta E = 1.14$ meV apart, two per level (spin up and down). Levels are numbered from 0 at the bottom. They fill from the bottom, so 6 electrons fill levels 0 to 2 and the 7th has to start level 3.

```python
n_fill = 6                                   # how many electrons to draw
fig, ax = plt.subplots(figsize=(5.2, 3.8))
for k in range(7):
    e = isl.eps[k]/meV                       # energy of level k in meV
    ax.hlines(e, 0.5, 2.5, color='k', lw=2)
    ax.text(2.65, e, f'level {k}', va='center', fontsize=8)
    for s, x in enumerate((1.0, 2.0)):       # two slots per level, spin up then spin down
        idx = 2*k + s                        # which electron (counting from 0) lives in this slot
        if idx < n_fill:   ax.text(x, e+0.05, '↑' if s == 0 else '↓', ha='center', va='bottom', fontsize=16, color='C0')
        elif idx == n_fill: ax.text(x, e+0.05, '↑', ha='center', va='bottom', fontsize=16, color='C3')   # the next one in, red
# arrow showing the spacing between level 0 and 1
ax.annotate('', xy=(0.25, isl.eps[1]/meV), xytext=(0.25, isl.eps[0]/meV), arrowprops=dict(arrowstyle='<->'))
ax.text(0.15, 0.5*(isl.eps[0]+isl.eps[1])/meV, 'ΔE', ha='right', va='center')
ax.set_xlim(-0.2, 3.5); ax.set_ylim(-1, 8.5); ax.set_xticks([]); ax.grid(False)
ax.set_ylabel('single-particle energy (meV)')
ax.set_title('6 electrons (blue) fill levels 0-2, the 7th (red) starts level 3', fontsize=9)
plt.show()
```

## 3. Energy of $N$ electrons against gate voltage

With $N$ electrons the energy is the filled ladder plus the charging energy,

$$E(N) = \sum_p n_p\epsilon_p + \frac{(N-Q)^2}{2C_\Sigma}, \qquad Q = q_0 + C_S V_S + C_D V_D + C_g V_G .$$

Each $N$ gives a parabola in $V_G$ and the island sits on the lowest one. Where two parabolas cross, adding an electron costs no energy. That crossing is the Coulomb peak, and with zero bias it's the only place current can flow.

```python
VGe = np.linspace(-300, 400, 800)*mV
Qe = isl.Q(0, 0, VGe)                                                # induced charge along the gate sweep, no bias
curves = {N: isl.E_ground(N, Qe)/meV for N in range(3, 10)}        # E(N) for a handful of N, in meV

fig, ax = plt.subplots(figsize=(8.8, 4))
for N, E in curves.items(): ax.plot(VGe/mV, E, label=f'N = {N}')
# thick grey line = lowest parabola at each V_G = what the island actually does
ax.plot(VGe/mV, np.min(list(curves.values()), axis=0), 'k', lw=3, alpha=.3, label='lowest = actual state')
for n in range(3, 10): ax.axvline(isl.V_peak(n)/mV, color='k', ls=':', lw=.9)   # dotted = Coulomb peaks
ax.set_xlim(-300, 400); ax.set_ylim(0, 35)
ax.set_xlabel('gate voltage $V_G$ (mV)  [$V_S = V_D = 0$]'); ax.set_ylabel('E(N) (meV)')
ax.set_title('Ground-state energy of N electrons, dotted lines = Coulomb peaks', fontsize=9)
ax.legend(fontsize=7, loc='center left', bbox_to_anchor=(1.01, .5)); plt.show()
```

## 4. Adding one electron

The cost of adding one more electron to an island holding $n$ is $\mu_{\rm add}(n) = E(n+1) - E(n)$. It drops linearly as $V_G$ goes up, and with source and drain at zero it hits zero at the Coulomb peaks.

The gap between neighbouring peaks is set by the addition energy $E_{\rm add}(n) = \mu_{\rm add}(n) - \mu_{\rm add}(n-1)$. It is $E_C$ when the new electron joins a spin partner in the same level, and $E_C + \Delta E$ when it has to start a new level. The peak spacing is $E_{\rm add}/(\alpha_g e)$, so it alternates between 140 and 154 mV.

```python
ns = np.arange(1, 13)
Eadd = np.array([isl.E_add(n)/meV for n in ns])

fig, ax = plt.subplots(figsize=(6, 3.4))
ax.bar(ns, Eadd, color=[EVEN_C if n % 2 == 0 else ODD_C for n in ns])
ax.axhline(isl.p.E_C/meV, color='k', ls='--', lw=1); ax.axhline((isl.p.E_C+isl.p.dE)/meV, color='k', ls='--', lw=1)
ax.text(12.6, isl.p.E_C/meV-0.5, '$E_C$', ha='right', va='top')
ax.text(12.6, (isl.p.E_C+isl.p.dE)/meV+0.1, '$E_C+\\Delta E$', ha='right', va='bottom')
ax.set_ylim(9, 14); ax.set_xticks(ns)
ax.set_xlabel('electrons already on the island, n'); ax.set_ylabel(r'$E_{add}(n)$ (meV)')
ax.set_title('addition energy alternates (even n starts a new level)', fontsize=9)
plt.show()
```

## 5. One diamond

Now put a bias $V_S$ on the source with $V_D = 0$. The source level moves to $\mu_S = -V_S$ (electrons are negative), the drain stays at $\mu_D = 0$, and the island levels shift too because $V_S$ changes $Q$. The island stays stuck at $n$ electrons (Coulomb blockade, no current) as long as

- the next electron can't get in: $\mu_{\rm add}(n) > \max(\mu_S, \mu_D)$
- the last electron can't get out: $\mu_{\rm add}(n-1) < \min(\mu_S, \mu_D)$

Plotting where this holds in $(V_G, V_S)$ gives a diamond, and the plot below comes straight from that rule.

```python
n0 = 6
vL, vR = isl.V_peak(n0-1)/mV, isl.V_peak(n0)/mV          # the two coulomb peaks either side of diamond n0 (mV)
VGd = np.linspace(vL-15, vR+15, 1500)*mV                 # fine grid around this one diamond
VSd = np.arange(-20, 20.001, 0.05)*mV
_, nmap = blockade_map(isl, VGd, VSd)                    # run the model, nmap says which n is stuck at each point

c = isl.C_S/isl.C_sigma
xa = vL + (1-c)*(vR-vL)                                  # top corner sits a fraction 1-c across
apex = isl.E_add(n0)/mV                                  # half height, E_add/e

fig, ax = plt.subplots(figsize=(8, 4.4))
inside = (nmap == n0).astype(float)                      # 1 inside diamond n0, 0 outside
ax.contourf(VGd/mV, VSd/mV, inside, levels=[0.5, 1.5], colors=[EVEN_C])
ax.contour(VGd/mV, VSd/mV, inside, levels=[0.5], colors='k', linewidths=1.2)
ax.plot(xa, apex, 'ro', ms=5); ax.text(xa+9, apex+3.5, f'apex = $E_{{add}}/e$ = {apex:.2f} mV', fontsize=8, color='r')
ax.annotate('', xy=(vL, -14.5), xytext=(vR, -14.5), arrowprops=dict(arrowstyle='<->'))     # width arrow
ax.text(0.5*(vL+vR), -13.7, f'width = $E_{{add}}/(\\alpha_g e)$ = {vR-vL:.1f} mV', ha='center', fontsize=8)
ax.text(vL+8, 12.0, f'edge slope +C_g/(C_Σ-C_S)\n= +{isl.C_g/(isl.C_sigma-isl.C_S):.2f}', fontsize=7.5)
ax.text(vR-30, 7.0, f'edge slope -C_g/C_S\n= {-isl.C_g/isl.C_S:.2f}', fontsize=7.5)
ax.text(0.5*(vL+vR), -6, f'BLOCKADE\n{n0} electrons, no current', ha='center', va='center', fontsize=9)
ax.text(vL-12, 17.3, 'current flows', fontsize=9)
ax.set_xlim(vL-15, vR+15); ax.set_ylim(-20, 20); ax.grid(False)
ax.set_xlabel('gate voltage $V_G$ (mV)'); ax.set_ylabel('source bias $V_S$ (mV)')
ax.set_title(f'The n = {n0} diamond (V_D = 0)', fontsize=10); plt.show()
```

## 6. The full diamond map

Same thing over the whole sweep. At each point of a grid in $(V_G, V_S)$, 1 mV by 0.1 mV steps, the model checks the blockade rule. White means current flows, light blue is blockade with even $n$ and dark blue is blockade with odd $n$, and the numbers are $n$.

```python
VG = np.arange(-600, 600.01, 1.0)*mV                    # same ranges as the thesis sweeps
VS = np.arange(-20, 20.01, 0.1)*mV
blockaded, n_map = blockade_map(isl, VG, VS)           # True where blockaded, n_map says which n

img = np.where(blockaded, 1 + (n_map % 2), 0)          # 0 = current flows, 1 = n even, 2 = n odd
fig, ax = plt.subplots(figsize=(10, 3.6))
ax.imshow(img, extent=[-600.5, 600.5, -20.05, 20.05], origin='lower', aspect='auto',
          cmap=ListedColormap(['white', EVEN_C, ODD_C]), vmin=0, vmax=2, interpolation='nearest')
ax.contour(VG/mV, VS/mV, blockaded.astype(float), levels=[0.5], colors='k', linewidths=.7)   # edges of the blockade
for n in range(1, 2*isl.p.K - 2):                      # write n in the middle of each diamond
    xc = 0.5*(isl.V_peak(n-1)+isl.V_peak(n))/mV
    if -600 < xc < 600: ax.text(xc, 0, str(n), ha='center', va='center', fontsize=9, color='k' if n % 2 == 0 else 'w')
ax.legend(handles=[Patch(fc='white', ec='k', label='current flows'), Patch(fc=EVEN_C, label='blockade, n even'), Patch(fc=ODD_C, label='blockade, n odd')],
          loc='upper right', fontsize=7, ncol=3)
ax.set_xlabel('gate voltage $V_G$ (mV)'); ax.set_ylabel('source bias $V_S$ (mV)'); ax.grid(False)
ax.set_title('Model at T = 0, number = electrons n', fontsize=10)
plt.show()
```

How this compares with thesis Fig. 8.4:

- The diamond shape is the same, with corners touching at $V_S = 0$.
- Apex height: the thesis has roughly 8 to 13 mV (10.6 to 13.1 mV for the central ones) and the model gives 11.2 and 12.3 mV.
- The thesis also has lines parallel to the diamond edges (excited states and finite temperature) and diagonal lines inside the blockade. The model has none of these (yet).

## What the model can't do yet

- It's at T = 0, so there is no thermal or tunnel broadening and everything is perfectly sharp.
- It only knows blockade or no blockade, not how much current flows.
- It uses ground states only, so no excited-state lines.
- The ladder is uniform and the capacitances are constant (Beenakker's assumptions), so real diamonds with irregular widths aren't reproduced.
- The source and drain are assumed symmetric ($c = 0.5$).
- There are no defects modelled yet.

# Diamonds at finite temperature

We need to add more levels as the electron can occupy levels above and below the ones in use. So we use n=48 and use 20 levels on either side which is more than the +-20mV bias can every reach.

T is the electron temperature, the fridge was at 7mK but we estimate 100mK as a guess.

```python
# an island with 48 levels, centred so that the levels in use have plenty of levels both above and below:
# electrons can then enter higher levels and leave from lower ones (the T = 0 island keeps 16 levels from n = 0)

isl_T = Island.centred(Params(K=48), n_ref=48)
n_shift = 48 - 6  # subtract from n so the diamond labels match the T = 0 plots

T_e = 100 * mK  # electron temperature (not measured; the fridge was at 7 mK)
VG = np.arange(-600, 600.01, 0.5) * mV  # same ranges as the thesis sweeps
VS = np.arange(-20, 20.01, 0.1) * mV

I_T, n_T = current_map(isl_T, VG, VS, kT=T_e)  # current and mean electron number
blockaded_T = np.abs(I_T) < 1e-6 * np.abs(I_T).max()  # "no current" = blockaded
blockaded_0, n_map_0 = blockade_map(isl_T, VG, VS)  # the T = 0 answer for the same island
print(f"points where {T_e / mK:.0f} mK and T = 0 agree on blockade: {100 * np.mean(blockaded_T == blockaded_0):.2f} %")
```

# What the model still cant do

- The equal level energy spacing means the Coulomb diamonds are still the same size and equally distanced.
- The assumption that the island relaxes into the ground state between tunnelling events may be wrong.
- We use the assumption that $\hbar \Gamma << kT$, they may be comparable.
- The small current inside the diamonds due to stray dopants giving a path for electrons to travel through is not modelled.
- Tunnel rates are not know so the current is in terms of $e\Gamma$
- Defects and stray dopants are the next stage of the project

### Test: Hofheinz's Fig. 5.4 (p. 70)

We run our code with his parameters from the Fig. 5.4 caption (p. 70) and Eq. 5.14 (p. 69), at his temperature of 1 K. His dot is metallic (p. 69), so we set $\Delta E = 0$ and use `metallic=True`.

He grounds the source and biases the drain (p. 27). We ground the drain and bias the source, so his trap on the source is on our drain, and his $V_d$ axis is our $V_S$.

What should match his figure:

- $\beta = 0.33$ and $\alpha_t = 0.015$ (he gives about 0.3 and 0.015 on p. 73).
- The two sets of diamonds, dotted for the dopant empty and dashed for it full, are $\delta V_G = 5.5$ mV apart.
- The zero-bias peaks are missing over about $\Delta V_G = 75$ mV around $V_G = 0$.
- The teeth sit on the positive slopes of the diamonds. This is his rule for a trap on the grounded lead (p. 73).

```python
# Test: Hofheinz's own parameters for Fig. 5.4 (thesis p. 70), converted to our convention.
# He grounds the source and biases the drain; we ground the drain and bias the source, so his "source" is our drain.
e_per_V = 1.0 / V                                      # he gives capacitances in e/V (p. 70); 1 e/V = 0.16 aF
p_H = Params(E_C=1 / (200 * e_per_V), alpha_g=60 / 200, c_S=70 / 200, dE=0.0, g=1, K=60)   # C = 200 e/V, metallic dot
isl_H = Island(p_H, q0=19.84)                          # q0 only slides the Coulomb peaks relative to the dopant
dop_H = Dopant(C_tm=1 * e_per_V, C_tG=0.045 * e_per_V, C_tS=0.0, C_tD=2 * e_per_V,          # his trap capacitances
               lead='D', Gamma_t=1e-3, Gamma_tm=1e-3, g_t=1)                               # Eq. 5.14 (p. 69), non-degenerate trap
dop_H = centre_dopant(isl_H, dop_H, 0.0)               # anomaly centred on V_G = 0, as in his figure
print(f"beta = {dop_H.beta:.3f},  alpha_t = {dop_H.alpha_G:.4f}")
print(f"Eq. 5.15: delta V_G = {dop_H.beta / isl_H.C_g / mV:.2f} mV,   "
      f"Eq. 5.16: Delta V_G = {dop_H.beta * (1 - dop_H.beta) * isl_H.p.E_C / dop_H.alpha_G / mV:.1f} mV")

VG_H = np.arange(-75, 75.01, 0.25) * mV                # his window
VS_H = np.arange(-7.5, 7.51, 0.05) * mV
I_H, occ_H, _, _, _ = dopant_map(isl_H, dop_H, VG_H, VS_H, kT=1000 * mK, metallic=True)    # his T = 1 K
```

```python
G_H = np.gradient(I_H, VS_H / mV, axis=0)                                # dI/dV_S
outline = [blockade_map(fixed_dopant_island(isl_H, dop_H, t), VG_H, VS_H)[1] for t in (0, 1)]   # T = 0 diamonds, empty / full
Q0, Qt0 = isl_H.Q(0.0, 0.0, VG_H), dop_H.Q_t(0.0, 0.0, VG_H)             # zero bias, for the energy panel
W = [np.array([dopant_energy(isl_H, dop_H, n, t, Q0, Qt0) for n in range(10, 32)]) / meV for t in (0, 1)]
W_ref = min(W[0].min(), W[1].min())

fig, ax = plt.subplots(3, 1, figsize=(6.6, 8.4), sharex=True)
for a in ax: a.tick_params(labelbottom=True)                         # show the V_G numbers on every panel
for t, col in ((0, 'b'), (1, 'k')):                                      # blue = dopant empty, black = full (as Hofheinz)
    for w in W[t]: ax[0].plot(VG_H / mV, w - W_ref, col, lw=0.4)
    ax[0].plot(VG_H / mV, W[t].min(axis=0) - W_ref, col, lw=2.5)          # thick: lowest island state for that dopant state
ax[0].set_ylim(-0.2, 2.2); ax[0].set_ylabel('E (meV, shifted)'); ax[0].grid(False)
ext = [VG_H[0] / mV, VG_H[-1] / mV, VS_H[0] / mV, VS_H[-1] / mV]
for a, data, lab in ((ax[1], occ_H, 'dopant occupation'), (ax[2], G_H / np.percentile(G_H, 99.5), r'$dI/dV_S$ (normalised)')):
    im = a.imshow(data, extent=ext, origin='lower', aspect='auto', cmap='Blues', vmin=0, vmax=1, interpolation='nearest')
    fig.colorbar(im, ax=a, label=lab)
    for nmap, ls in zip(outline, (':', '--')):                           # dotted = dopant empty, dashed = dopant full
        a.contour(VG_H / mV, VS_H / mV, (nmap >= 0).astype(float), levels=[0.5], colors='k', linewidths=0.6, linestyles=ls)
    a.set_ylabel('source bias $V_S$ (mV)'); a.grid(False)
for a in ax: a.set_xlabel('gate voltage $V_G$ (mV)')                 # x label on every panel
ax[0].set_title("Our model with Hofheinz's Fig. 5.4 parameters (T = 1 K)", fontsize=10)
fig.tight_layout()
plt.show()
```

### $\beta$ for island 1 from Coulomb's law

$\beta = U(d)/E_C$ (section 6.4), where $U(d)$ is the Coulomb energy between an island electron and a dopant electron a distance $d$ apart in silicon. The table gives $\beta$ and the shift of the diamonds when the dopant is full, $\delta V_G = \beta e/C_g$. These are upper estimates, because screening by the leads and gates is ignored.

```python
# beta from Coulomb's law for a dopant a distance d from island 1: beta = U(d) / E_C
eps_r = ElectrostaticsParams().eps_r_Si
print(f"C_sigma = {isl_T.C_sigma / aF:.2f} aF,  e/C_g = {1 / isl_T.C_g / mV:.0f} mV")
print(" d (nm)   U(d) (meV)   beta    delta V_G = beta e/C_g (mV)")
for d in (20, 30, 50, 100):
    U = 1.0 / (eps_r * d * nm)                         # e^2 / (4 pi eps0 eps_r d), with e = 4 pi eps0 = 1
    beta = U / isl_T.p.E_C
    print(f"{d:6d}   {U / meV:9.2f}   {beta:6.3f}   {beta / isl_T.C_g / mV:8.1f}")
```

### Island 1 with one stray dopant: the choices (from `DopantParams`)

| Parameter | Value | Why |

| $\alpha_G$ | 0.004 | not known. In Hofheinz's device the trap's gate lever arm is $0.015/0.3 = 0.05$ of the island's (pp. 70, 73). The same fraction of island 1's 0.08 gives 0.004 |

### What this step does not include yet

- The capacitances are estimates. $\beta$ comes from Coulomb's law without screening, $\alpha_G$ is scaled from Hofheinz's device, and the rest goes to one lead. The electrostatics stage may compute all of them.
- The capacitances are constant. A barrier capacitance can change when a dopant fills (Hofheinz pages 89–90).
- The dopant cannot tunnel to the island. (Eq. 2.21, page 25).
- The dopant does not change the island's tunnel rates $\Gamma_S$ and $\Gamma_D$ (Hofheinz needed this for his Fig. 5.9, p. 80).
- There is only one dopant, so there are no rearrangements between several dopants (Hofheinz pp. 75–76).
- The dopant is assumed slow, $\Gamma_t \ll \Gamma$. The error is of order $\Gamma_t/\Gamma$.
- The dopant's level (set by $V_c$) and the electron temperature are not known.

### A dopant near the source

The section 7 dopant, but near the source. It tunnels to the source, fills against the source's Fermi level $\mu_S = -V_S$, and the capacitance not to the island or the gate goes to the source, $\alpha_S = 1 - \beta - \alpha_G$.

The cell first checks the zero-bias row. Both leads have the same Fermi level there, so which lead the dopant uses cannot matter.

The dopant's level moves with the voltages as
$$d\mu_t = -(\beta\alpha_g + \alpha_G)\,dV_G - (\beta c_S + \alpha_S)\,dV_S .$$
It fills where its level crosses its own lead's Fermi level, so the line where it fills has slope

- near the drain ($\alpha_S = 0$, $\mu_D = 0$): $dV_S/dV_G = -(\beta\alpha_g + \alpha_G)/(\beta c_S) = -0.20$,
- near the source ($\alpha_S = 0.776$, $\mu_S = -V_S$): $dV_S/dV_G = -(\beta\alpha_g + \alpha_G)/(\beta c_S + \alpha_S - 1) = +0.19$.

So at zero bias nothing changes, with a bias the map is close to the section 7 map turned upside down, and the teeth move from the positive slopes of the diamonds to the negative slopes.

```python
# A dopant near the source: the section 7 dopant, but it tunnels to the source, uses the source's Fermi level,
# and the capacitance that is not to the island or the gate goes to the source (DopantParams, lead = 'S')
dp = DopantParams()
dop_S = centre_dopant(isl_T, Dopant.from_coulomb(isl_T, d=dp.d, alpha_G=dp.alpha_G, lead='S', g_t=dp.g_t,
                                                 Gamma_t=dp.Gamma_t), dp.V_c)
I_Sd, occ_Sd, _, _, _ = dopant_map(isl_T, dop_S, VG5, VS5, kT=T_e)    # the full map, about 1 minute

# Check: at zero bias both leads have the same Fermi level, so it must not matter which lead the dopant uses
j0 = np.argmin(np.abs(VS5))
_, occ_D0, _, _, _ = dopant_map(isl_T, dop_T, VG5, VS5[j0:j0 + 1], kT=T_e)
print(f"zero bias, source dopant against drain dopant: largest difference in the occupation {np.abs(occ_Sd[j0] - occ_D0[0]).max():.1e}")

dIdV_Sd = np.gradient(I_Sd, VS5 / mV, axis=0)
real_Sd = np.abs(I_Sd) < 1e-6 * np.abs(I_Sd).max()                    # the real diamonds
outline_Sd = [blockade_map(fixed_dopant_island(isl_T, dop_S, t), VG5, VS5)[1] for t in (0, 1)]
fig = plt.figure(figsize=(10, 9.5))
gs = fig.add_gridspec(3, 2, width_ratios=[1, 0.025], wspace=0.03, hspace=0.38)
ax = [fig.add_subplot(gs[i, 0]) for i in range(3)]
cax = [fig.add_subplot(gs[i, 1]) for i in range(3)]
ext = [VG5[0] / mV, VG5[-1] / mV, VS5[0] / mV, VS5[-1] / mV]
maps = ((occ_Sd, 1, 'dopant occupation'), (np.abs(I_Sd), np.abs(I_Sd).max(), r'$|I|$ ($e\Gamma$)'),
        (dIdV_Sd / np.percentile(dIdV_Sd, 99.5), 1, r'$dI/dV_S$ (normalised)'))
for a, c, (data, vmax, lab) in zip(ax, cax, maps):
    im = a.imshow(data, extent=ext, origin='lower', aspect='auto', cmap='Blues', vmin=0, vmax=vmax, interpolation='antialiased')
    fig.colorbar(im, cax=c, label=lab)
    for nmap, ls in zip(outline_Sd, (':', '--')):                      # T = 0 diamonds, dopant empty / full
        a.contour(VG5 / mV, VS5 / mV, (nmap >= 0).astype(float), levels=[0.5], colors='0.35', linewidths=0.6, linestyles=ls)
    a.contour(VG5 / mV, VS5 / mV, real_Sd.astype(float), levels=[0.5], colors='k', linewidths=1.1)   # real diamonds
    for n in range(1, 2 * isl_T.p.K - 2):                              # same diamond numbers as section 7
        xc = 0.5 * (isl_T.V_peak(n - 1) + isl_T.V_peak(n)) / mV
        if -600 < xc < 600: a.text(xc, 0, str(n - n_shift), ha='center', va='center', fontsize=8)
    a.set_xlim(ext[0], ext[1])
    a.set_ylabel('source bias $V_S$ (mV)')
    a.set_xlabel('gate voltage $V_G$ (mV)')
    a.grid(False)
ax[0].set_title('Dopant near the source. Solid black: real diamonds, dotted / dashed: T = 0, dopant empty / full', fontsize=9)
plt.show()
```

### Five dopants

From `ManyDopantParams`. Each dopant $i$ is placed at random 40 to 100 nm from the island and uses the lead on its side. It gets

### The maps for five dopants

Top to bottom: the mean number of full dopants $\sum_i p_i$, $|I|$, $dI/dV_S$ and $dI/dV_G$. The solid black lines are the real diamonds. The dotted lines are the diamonds of the plain island with every dopant empty.

- Each dopant fills somewhere in the map, so the number of full dopants rises from 0 on the left to 2 or 3 on the right. Several dopants are full at once.
- Wherever dopant $i$ fills, the diamonds jump to the right by $\delta V_{G,i} = \beta_i e/C_g$, between 17 and 32 mV.
- $dI/dV_G$ is red where the current rises with $V_G$ and blue where it falls, so each peak shows as a red line then a blue line.
- When dopant $j$ fills, dopant $i$'s level rises by $\beta_i\beta_jE_C + U_{ij}$, so where one fills depends on which others are full. The map is not just five copies of the section 7 map.

```python
# The maps: number of full dopants, |I|, dI/dV_S and dI/dV_G, with the real diamonds
dIdVS8 = np.gradient(I8, VS8 / mV, axis=0)
dIdVG8 = np.gradient(I8, VG8 / mV, axis=1)
real8 = np.abs(I8) < 1e-6 * np.abs(I8).max()
empty8 = blockade_map(isl_T, VG8, VS8)[1]                              # T = 0 diamonds with every dopant empty
fig = plt.figure(figsize=(10, 12.5))
gs = fig.add_gridspec(4, 2, width_ratios=[1, 0.025], wspace=0.03, hspace=0.4)
ax = [fig.add_subplot(gs[i, 0]) for i in range(4)]
cax = [fig.add_subplot(gs[i, 1]) for i in range(4)]
ext = [VG8[0] / mV, VG8[-1] / mV, VS8[0] / mV, VS8[-1] / mV]
gmax = np.percentile(np.abs(dIdVG8), 99.5)
maps = ((nf8, 'Blues', 0, ds.N, 'number of full dopants'), (np.abs(I8), 'Blues', 0, np.abs(I8).max(), r'$|I|$ ($e\Gamma$)'),
        (dIdVS8 / np.percentile(dIdVS8, 99.5), 'Blues', 0, 1, r'$dI/dV_S$ (normalised)'),
        (dIdVG8 / gmax, 'RdBu_r', -1, 1, r'$dI/dV_G$ (normalised)'))
for a, c, (data, cm, lo, hi, lab) in zip(ax, cax, maps):
    im = a.imshow(data, extent=ext, origin='lower', aspect='auto', cmap=cm, vmin=lo, vmax=hi, interpolation='antialiased')
    fig.colorbar(im, cax=c, label=lab)
    a.contour(VG8 / mV, VS8 / mV, (empty8 >= 0).astype(float), levels=[0.5], colors='0.35', linewidths=0.6, linestyles=':')
    a.contour(VG8 / mV, VS8 / mV, real8.astype(float), levels=[0.5], colors='k', linewidths=1.0)
    a.set_xlim(ext[0], ext[1])
    a.set_ylabel('$V_S$ (mV)')
    a.set_xlabel('gate voltage $V_G$ (mV)')
    a.grid(False)
ax[0].set_title(f'{ds.N} stray dopants. Solid black: real diamonds, dotted: T = 0 diamonds with every dopant empty', fontsize=9)
plt.show()
```

### The point the dopants are tuned at

The tuning (functions notebook, 7.5) is done at the point a noise measurement would sit on: a source bias of 0.1 mV, and the steepest point on the rising side of the peak between diamonds 6 and 7, with every dopant empty. It is at $V_G = 76.457$ mV, just left of the peak at 77.1 mV. It is the orange cross on the maps below.

```python
# The point the dopants are tuned at: small bias, steepest point on the rising side of the 6-7 peak (every dopant empty)
opp = OperatingParams()
V_pk = isl_T.V_peak(opp.n_peak + n_shift)                              # the peak between diamonds 6 and 7
VG_op = np.arange(V_pk / mV - 5, V_pk / mV + 5, 0.002) * mV            # 2 microvolt steps around it
I_e, _ = current_map(isl_T, VG_op, np.array([opp.V_S_op]), kT=T_e)     # the current along the gate there
k0, slope0 = flank_point(VG_op, I_e[0])                                # its steepest rising point
V_op = VG_op[k0]
print(f"tuning point: V_S = {opp.V_S_op / mV:.1f} mV, V_G = {V_op / mV:.3f} mV, current {I_e[0, k0]:.3f} e Gamma "
      f"(the peak is at {V_pk / mV:.1f} mV)")
```

### Every dopant tuned to rate in = rate out, interactions off

The five dopants above, with each one's level changed so that, on its own, $K_\text{in} = K_\text{out}$ (so $p = 1/2$) at the tuning point (`balance_each`). Their positions, couplings and tunnel rates are unchanged. The maps are drawn with the interactions off (`interacting=False`), so each dopant keeps the rates it has on its own.

- At the tuning point every dopant is full half the time (the printout).
- Left of the tuning point, at small bias, every dopant is empty, so the diamonds are those of the plain island (solid and dotted lines agree).
- Right of it the dopants fill one after another, the number of full dopants rises to 5, and every filling moves the diamonds to the right by 17 to 32 mV, so the diamonds break up.
- $p = 1/2$ holds only at the tuning point. Whether a dopant is full depends on the gate voltage, so it cannot be $1/2$ across a whole map.

```python
# Every dopant tuned so that, on its own, K_in = K_out (p = 1/2) at the tuning point. Map with the interactions off.
dsN = balance_each(isl_T, ds, V_op, opp.V_S_op, kT=T_e)               # same positions, couplings and rates, new levels
B = configs(dsN.N)
Wn, _ = point_chain(isl_T, dsN, V_op, opp.V_S_op, kT=T_e, interacting=False)
print(f"interactions off, at the tuning point: p_i = {np.round(B.T @ stationary_point(Wn), 3)}")
I8n, nf8n, p8n = many_dopant_map(isl_T, dsN, VG8, VS8, kT=T_e, interacting=False)    # about 3 minutes

def plot_maps(I_m, nf_m, title):
    """The four maps, laid out as for the random levels above, with an orange cross at the tuning point."""
    dIdVS_m = np.gradient(I_m, VS8 / mV, axis=0)
    dIdVG_m = np.gradient(I_m, VG8 / mV, axis=1)
    real_m = np.abs(I_m) < 1e-6 * np.abs(I_m).max()
    fig = plt.figure(figsize=(10, 12.5))
    gs = fig.add_gridspec(4, 2, width_ratios=[1, 0.025], wspace=0.03, hspace=0.4)
    ax = [fig.add_subplot(gs[i, 0]) for i in range(4)]
    cax = [fig.add_subplot(gs[i, 1]) for i in range(4)]
    ext = [VG8[0] / mV, VG8[-1] / mV, VS8[0] / mV, VS8[-1] / mV]
    gmax = np.percentile(np.abs(dIdVG_m), 99.5)
    maps = ((nf_m, 'Blues', 0, ds.N, 'number of full dopants'), (np.abs(I_m), 'Blues', 0, np.abs(I_m).max(), r'$|I|$ ($e\Gamma$)'),
            (dIdVS_m / np.percentile(dIdVS_m, 99.5), 'Blues', 0, 1, r'$dI/dV_S$ (normalised)'),
            (dIdVG_m / gmax, 'RdBu_r', -1, 1, r'$dI/dV_G$ (normalised)'))
    for a, c, (data, cm, lo, hi, lab) in zip(ax, cax, maps):
        im = a.imshow(data, extent=ext, origin='lower', aspect='auto', cmap=cm, vmin=lo, vmax=hi, interpolation='antialiased')
        fig.colorbar(im, cax=c, label=lab)
        a.contour(VG8 / mV, VS8 / mV, (empty8 >= 0).astype(float), levels=[0.5], colors='0.35', linewidths=0.6, linestyles=':')
        a.contour(VG8 / mV, VS8 / mV, real_m.astype(float), levels=[0.5], colors='k', linewidths=1.0)
        a.plot([V_op / mV], [opp.V_S_op / mV], 'x', color='tab:orange', ms=9, mew=2)
        a.set_xlim(ext[0], ext[1])
        a.set_ylabel('$V_S$ (mV)')
        a.set_xlabel('gate voltage $V_G$ (mV)')
        a.grid(False)
    ax[0].set_title(title + '. Orange cross: the tuning point', fontsize=9)
    plt.show()

plot_maps(I8n, nf8n, 'every dopant tuned to rate in = rate out, interactions off')
```

### The same tuned dopants, interactions on

The same levels as above, now with the interactions on: a full dopant raises the others' levels by $\beta_i\beta_jE_C + U_{ij}$, 100 to 600 times $kT$.

- At the tuning point each dopant is now full only about 1/6 of the time instead of 1/2 (the printout). One full dopant pushes the others' levels up, so they take turns instead of filling together.
- Right of the tuning point only one or two dopants are full anywhere on the map, instead of up to five, so the diamonds are broken up much less.

So tuning each dopant on its own does not give rate in = rate out once they interact. The next cell tries to tune all five together.

```python
# The same tuned levels, with the interactions on: the dopants push on each other again
Wt, _ = point_chain(isl_T, dsN, V_op, opp.V_S_op, kT=T_e, interacting=True)
print(f"interactions on, at the tuning point: p_i = {np.round(B.T @ stationary_point(Wt), 3)}  (0.5 each with them off)")
I8t, nf8t, p8t = many_dopant_map(isl_T, dsN, VG8, VS8, kT=T_e)        # about 3 minutes
plot_maps(I8t, nf8t, 'the same tuned levels, interactions on')
```

### Why the tuning cannot be done with the interactions on

`balance_all` (functions notebook, 7.6) changes each dopant's level in turn until its average rate in equals its average rate out with the interactions on, repeated for 10 rounds.

- It does not get there: after 10 rounds only the last dopant is at $p = 1/2$, and the others are at 0.04 to 0.26.
- Exactly one dopant is full at all times.
- The average rates have collapsed to $10^{-17}$ to $10^{-21}\,\Gamma$, so the dopants hardly switch at all.

Setting one dopant pushes the others' levels by hundreds of $kT$, so they end up stuck. With couplings this large, rate in = rate out can only be set for every dopant if they do not interact.

```python
# Can every dopant be tuned to rate in = rate out at once, with the interactions on? Each dopant's level is changed in turn
# until its average K_in equals its average K_out (p_i = 1/2), repeated over all five, 10 rounds (about 2 minutes)
dsJ, n_sw = balance_all(isl_T, dsN, V_op, opp.V_S_op, kT=T_e)
WJ, _ = point_chain(isl_T, dsJ, V_op, opp.V_S_op, kT=T_e, interacting=True)
PJ = stationary_point(WJ)
print(f"after {n_sw} rounds: p_i = {np.round(B.T @ PJ, 3)}  (all 0.5 would mean every dopant balanced)")
print("fraction of time with 0, 1, 2, ... dopants full:", [round(float(PJ[B.sum(axis=1) == n].sum()), 3) for n in range(dsJ.N + 1)])
print("dopant   average K_in   average K_out")
for i in range(dsJ.N):
    emp = np.where(B[:, i] == 0)[0]
    ful = np.where(B[:, i] == 1)[0]
    k_in = (PJ[emp] * WJ[emp, emp ^ (1 << i)]).sum() / PJ[emp].sum()
    k_out = (PJ[ful] * WJ[ful, ful ^ (1 << i)]).sum() / PJ[ful].sum()
    print(f"  {i}      {k_in:9.1e}      {k_out:9.1e}")
```

### What this does not include yet

- The number of dopants, where they are and their tunnel rates are choices. The tunnel rates are spaced evenly in log, not worked out from each dopant's distance to its lead.
- $U_{ij}$ and $\beta_i$ are bare Coulomb values. The leads and gates would screen them, so the shifts of 17 to 32 mV are probably too large.
- The tuning to rate in = rate out only holds at one point, the tuning point.
